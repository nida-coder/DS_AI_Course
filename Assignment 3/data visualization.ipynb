```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt


# Load your dataset
df = pd.read_csv('cleaned customer churn dataset.csv')

# View first few rows
df.head()
```

```python
sns.countplot(x='Churn', data=df, palette='coolwarm')
plt.title('Number of Customers by Churn Status', fontsize=14)
plt.xlabel('Churn')
plt.ylabel('Number of Customers')
plt.show()
```

```python

# 3️⃣ Scatter Plot – Total Spend vs Tenure
sns.scatterplot(x='tenure', y='total spend', hue='churn', data=df, palette='coolwarm')
plt.title('Total Spend vs Tenure by Churn Status', fontsize=14)
plt.xlabel('Tenure (months)')
plt.ylabel('Total Spend')
plt.show()
```

```python
# 4️⃣ Box Plot – Total Spend by Churn (Aesthetic Version)
sns.set_style('whitegrid')  # clean background
sns.boxplot(
    x='Churn',
    y='Total Spend',
    data=df,
    palette=['#69b3a2', '#ff9999'],  # pleasant colors
    width=0.5,
    linewidth=1.5
)
plt.title('Total Spend by Churn Status', fontsize=14, fontweight='bold')
plt.xlabel('Churn', fontsize=12)
plt.ylabel('Total Spend', fontsize=12)
plt.show()
```

```python
# 5️⃣ Count Plot – Contract Length vs Churn
sns.set_style('whitegrid')  # clean and modern style

sns.countplot(
    x='Contract Length',     # Correct column name
    hue='Churn',             # Correct column name
    data=df,
    palette=['#66b3ff', '#ff6666'],  # Soft and balanced colors
    edgecolor='black'
)

plt.title('Contract Length vs Churn', fontsize=14, fontweight='bold')
plt.xlabel('Contract Length', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.legend(title='Churn', loc='upper right')
plt.show()
```
